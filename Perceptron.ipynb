# Perceptron do Zero com Python e NumPy

Este notebook demonstra o Perceptron binário de Rosenblatt, com os dados das portas lógicas AND, OR e XOR. O código do classificador está em `src/perceptron.py`.

**Ambiente:** Jupyter local ou Google Colab. No Colab, envie a pasta do projeto ou clone o repositório antes de executar as células. Não há treinamento prévio embutido: os resultados serão gerados ao executar as células.


## 1. Conceito

O Perceptron é um classificador linear. Calcula $z = \mathbf{w}^T\mathbf{x} + b$ e prevê uma de duas classes, aqui `-1` ou `+1`.

Quando uma amostra é classificada incorretamente (ou está na fronteira), atualiza pesos e viés com $\mathbf{w} \leftarrow \mathbf{w} + \eta y\mathbf{x}$ e $b \leftarrow b + \eta y$, onde $\eta$ é a taxa de aprendizado.


```python
import numpy as np
import matplotlib.pyplot as plt
from src.perceptron import Perceptron

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_and = np.array([-1, -1, -1, 1])
y_or = np.array([-1, 1, 1, 1])
y_xor = np.array([-1, 1, 1, -1])
```

## 2. Treinamento das portas AND e OR

Para cada porta, o código imprime o número de épocas, os pesos, o viés e a acurácia **nos dados de treinamento**.


```python
modelos = {}
for nome, y in [("AND", y_and), ("OR", y_or)]:
    modelo = Perceptron(taxa_aprendizado=0.1, epocas=100).ajustar(X, y)
    modelos[nome] = modelo
    print(f"{nome}: acurácia de treino = {modelo.acuracia(X, y):.2%}; épocas = {len(modelo.erros_por_epoca_)}")
    print(f"pesos = {modelo.pesos_}; viés = {modelo.vies_:.3f}; previsão = {modelo.prever(X)}")
```

## 3. Visualização dos erros por época


```python
fig, ax = plt.subplots(figsize=(7, 4))
for nome, modelo in modelos.items():
    ax.plot(range(1, len(modelo.erros_por_epoca_) + 1), modelo.erros_por_epoca_, marker='o', label=nome)
ax.set(xlabel="Época", ylabel="Atualizações (erros)", title="Treinamento do Perceptron")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

## 4. Fronteiras de decisão

Como as entradas possuem duas variáveis, a fronteira pode ser representada em um plano.


```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
xx, yy = np.meshgrid(np.linspace(-0.5, 1.5, 200), np.linspace(-0.5, 1.5, 200))
malha = np.c_[xx.ravel(), yy.ravel()]
for ax, (nome, y) in zip(axes, [("AND", y_and), ("OR", y_or)]):
    pred = modelos[nome].prever(malha).reshape(xx.shape)
    ax.contourf(xx, yy, pred, alpha=0.2, levels=[-1.5, 0, 1.5])
    ax.scatter(X[:, 0], X[:, 1], c=y, edgecolors='black', s=100)
    ax.set(xlabel="Entrada 1", ylabel="Entrada 2", title=f"Fronteira de decisão: {nome}", xlim=(-0.5, 1.5), ylim=(-0.5, 1.5))
plt.tight_layout()
plt.show()
```

## 5. Limitação: XOR não é linearmente separável

Não existe uma única reta que separe corretamente todos os pontos de XOR. O Perceptron simples, portanto, não consegue representar essa função com classificação perfeita de todos os quatro exemplos.


```python
modelo_xor = Perceptron(taxa_aprendizado=0.1, epocas=100).ajustar(X, y_xor)
print("XOR, previsões:", modelo_xor.prever(X))
print(f"XOR, acurácia no conjunto de treino: {modelo_xor.acuracia(X, y_xor):.2%}")
```

## 6. Conclusão

O Perceptron aprende exemplos linearmente separáveis, como AND e OR. A limitação com XOR motiva modelos com camadas ocultas e funções de ativação não lineares. Estes exemplos são didáticos e não substituem uma avaliação em dados reais com divisão de treino e teste.
